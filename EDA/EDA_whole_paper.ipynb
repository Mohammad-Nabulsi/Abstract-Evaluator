```python
first_reviews = df.iloc[0]["reviews"]

# convert reviews to strings for uniqueness check
unique_reviews = list(set(map(str, first_reviews)))

print(f"Total reviews: {len(first_reviews)}")
print(f"Unique reviews: {len(unique_reviews)}")

for i, review in enumerate(unique_reviews, 1):
    print(f"\n--- Unique Review {i} ---")
    print(review)
```

```python
# One row per review
df_reviews = (
    df[["paper_id", "reviews"]]
    .explode("reviews")
    .dropna(subset=["reviews"])
    .rename(columns={ "reviews": "review"})
    .copy()
)

# review number within each paper
df_reviews["review_id"] = df_reviews.groupby("paper_id").cumcount() + 1

# reorder columns
df_reviews = df_reviews[["paper_id", "review_id", "review"]]
print('==='*10)
print(df_reviews.shape)
print('==='*10)
df_reviews.head()
```

```python
# One row per unique review per paper
df_reviews = (
    df[["paper_id", "reviews"]]
    .explode("reviews")
    .dropna(subset=["reviews"])
    .rename(columns={"reviews": "review"})
    .copy()
)

# convert reviews to string temporarily for duplicate detection
df_reviews["review_str"] = df_reviews["review"].astype(str)

# remove duplicate reviews within the same paper
df_reviews = df_reviews.drop_duplicates(
    subset=["paper_id", "review_str"]
)

# regenerate sequential review ids per paper
df_reviews["review_id"] = (
    df_reviews.groupby("paper_id").cumcount() + 1
)

# final columns
df_reviews = df_reviews[
    ["paper_id", "review_id", "review"]
].reset_index(drop=True)
print('==='*10)
print(df_reviews.shape)
print('==='*10)
df_reviews.head()
```

```python
# Keep all original columns while creating one row per unique review

df_reviews = (
    df
    .explode("reviews")
    .dropna(subset=["reviews"])
    .rename(columns={"reviews": "review"})
    .copy()
)

# temporary string version for duplicate checking
df_reviews["review_str"] = df_reviews["review"].astype(str)

# remove duplicate reviews within the same paper
df_reviews = df_reviews.drop_duplicates(
    subset=["paper_id", "review_str"]
)

# generate sequential review ids per paper
df_reviews["review_id"] = (
    df_reviews.groupby("paper_id").cumcount() + 1
)

# remove helper column
df_reviews = df_reviews.drop(columns=["review_str"])

df_reviews.head()
```
